```python
import json

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Загрузка данных

```python
# загрузка events.json
with open("events.json", encoding="utf-8") as f:
    data = json.load(f)

events_df = pd.DataFrame(data["events"])
print(events_df.shape)
events_df.head()
```

## Распределение событий по типам

```python
# распределение по signature
signature_counts = events_df["signature"].value_counts()
print("Распределение событий по типам:")
print(signature_counts)
print(f"\nВсего типов событий: {len(signature_counts)}")
```

## График

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=events_df, y="signature", hue="signature",
              order=signature_counts.index, legend=False, palette="viridis")
plt.title("Распределение типов событий информационной безопасности")
plt.xlabel("Количество событий")
plt.ylabel("Тип события (signature)")
plt.tight_layout()
plt.savefig("events_distribution.png")
plt.show()
```

## Срабатывания по категориям

```python
# первое слово сигнатуры — категория
category_counts = {}
for signature, count in signature_counts.items():
    category = signature.split()[0]
    category_counts[category] = category_counts.get(category, 0) + count

print(category_counts)
```
